# Convert 2 JJ to 1 JJ, remember to double Cj. Reduce Lj by half.

```python
# Defining the Junction in pyEPR
pinfo.junctions['j1'] = {'Lj_variable' : 'Q0_bottom_Lj',        # Inductance Variable Name
                         'rect'        : 'JJ_rect_Lj_Q0_bottom_rect_jj_L',  # Name of the rectangle geometry where Lumped RLC is specified
                         'line'        : 'JJ_Lj_Q0_bottom_rect_jj_L_',   # HFSS Polyline that spans the length of the rectangle
                         'length'      : epr.parse_units('0.06mm'),
                         'Cj_variable' : 'Q0_bottom_Cj'}  
pinfo.junctions['j2'] = {'Lj_variable' : 'Q0_top_Lj',        # Inductance Variable Name
                         'rect'        : 'JJ_rect_Lj_Q0_top_rect_jj_L',  # Name of the rectangle geometry where Lumped RLC is specified
                         'line'        : 'JJ_Lj_Q0_top_rect_jj_L_',   # HFSS Polyline that spans the length of the rectangle
                         'length'      : epr.parse_units('0.06mm'),
                         'Cj_variable' : 'Q0_top_Cj'}  

pinfo.junctions['j3'] = {'Lj_variable' : 'Q0_bottom_Lj',        # Inductance Variable Name
                         'rect'        : 'JJ_rect_Lj_Q0_bottom_rect_jj_L',  # Name of the rectangle geometry where Lumped RLC is specified
                         'line'        : 'JJ_Lj_Q0_bottom_rect_jj_L_',   # HFSS Polyline that spans the length of the rectangle
                         'length'      : epr.parse_units('0.06mm'),
                         'Cj_variable' : 'Q0_bottom_Cj'}  
pinfo.validate_junction_info()  # Raises an error if something is wrong with junction definition
```

```python
# # Simulation Parameters

pinfo.setup.min_pass = 6
pinfo.setup.passes = 15    # Maximum Number of passes
pinfo.setup.n_modes = 9     # Specifying number of modes for eigenmode analysis
pinfo.setup.delta_f = 1
pinfo.setup.min_freq = '1.0GHz'
```

### Run Eigenmode Simulations

```python
pinfo.setup.analyze()
```
